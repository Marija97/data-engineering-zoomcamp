```python
import pandas as pd
```

```python
# Prepare data url
prefix = 'https://github.com/DataTalksClub/nyc-tlc-data/releases/download/yellow/'
url = f'{prefix}/yellow_tripdata_2021-01.csv.gz'
url
```

```python
# Specify the data types
dtype = {
    "VendorID": "Int64",
    "passenger_count": "Int64",
    "trip_distance": "float64",
    "RatecodeID": "Int64",
    "store_and_fwd_flag": "string",
    "PULocationID": "Int64",
    "DOLocationID": "Int64",
    "payment_type": "Int64",
    "fare_amount": "float64",
    "extra": "float64",
    "mta_tax": "float64",
    "tip_amount": "float64",
    "tolls_amount": "float64",
    "improvement_surcharge": "float64",
    "total_amount": "float64",
    "congestion_surcharge": "float64"
}

parse_dates = [
    "tpep_pickup_datetime",
    "tpep_dropoff_datetime"
]
```

```python
# Read all of the available data with parsed types
df = pd.read_csv(url, dtype=dtype, parse_dates=parse_dates)
```

```python
# Create Database Connection
from sqlalchemy import create_engine
engine = create_engine('postgresql+psycopg://root:root@localhost:5432/ny_taxi')
```

```python
engine
```

```python
# Get DDL Schema
print(pd.io.sql.get_schema(df, name='yellow_taxi_data', con=engine))
```

```python
# Create the Table
# Note: head(n=0) makes sure we only create the table, we don't add any data yet.
# if the table exists, this line drops it and creates new table with no data
df.head(n=0).to_sql(name='yellow_taxi_data', con=engine, if_exists='replace')
```

```python
# Collect data for ingestion into an iterator of chunks
df_iter = pd.read_csv(
    prefix + 'yellow_tripdata_2021-01.csv.gz',
    dtype=dtype,
    parse_dates=parse_dates,
    iterator=True,
    chunksize=100000
)
```

```python
!uv add tqdm
```

```python
# Ingest data in chunks, with a progress bar from tqdm
from tqdm.auto import tqdm

for df_chunk in tqdm(df_iter):
    df_chunk.to_sql(name="yellow_taxi_data", con=engine, if_exists="append")
    print("Inserted chunk:", len(df_chunk))
```
